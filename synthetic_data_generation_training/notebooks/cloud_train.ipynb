## 8. Next Steps <a class="anchor" id="head-8"></a>

#### Generating Synthetic Data for your use case:

* Additional changes in the Domain Randomization under the Synthetic Data Generation script (`palletjack_sdg/standalone_palletjack_sdg.py`
* Add additional objects of interest in the scene (add `forklifts`, `ladders` etc.) to generate data
* [different](https://docs.nvidia.com/tao/tao-toolkit/text/tao_toolkit_quick_start_guide.html#downloading-the-models) models to be used for training with TAO (for object detection `YOLO`, `SSD`, `EfficientDet` can be used) 
* Replicator provides Semantic Segmentation, Instance Segmentation, Depth and various other ground truth annotations along with RGB. Custom ground truth annotator can be written (eg: Pose Estimation: [sample](https://docs.omniverse.nvidia.com/isaacsim/latest/tutorial_replicator_offline_pose_estimation.html). These can be used for training a model of diferent framework and choice
* The option of using `Synthetic + Real` data for training a network. Can be particularly useful for generating more data around particular corner cases


#### Deploying Trained Models:

* The model can be further optimized for deployment with the help of Pruning and QAT.
* TAO models can directly be deployed on Jetson with Isaac ROS or Deepstream which ensures end-to-end pipeline being optimized: data acquisition -> model inference -> results
